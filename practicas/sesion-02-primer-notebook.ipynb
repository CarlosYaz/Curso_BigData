Tarea 3 · Las cámaras de frío, con variables
Cada tienda tiene 3 cámaras de frío, y cada cámara envía su temperatura 12 veces por hora, las 24 horas del día.

```python
tiendas = 120
camaras_por_tienda = 3
lecturas_por_hora = 12
lecturas_al_dia = ((lecturas_por_hora * 24)*camaras_por_tienda)*(tiendas)
lecturas_al_anio = lecturas_al_dia * 365
print(f"Lecturas al día: {lecturas_al_dia}")
print(f"Lecturas al año: {lecturas_al_anio}")
```

Tarea 5 · Los clics de la web y la app
La web y la app de Colmena Market reciben unas 250.000 visitas al día. En cada visita se registran unos 12 clics, y cada clic ocupa unos 220 bytes.

Calcula, con variables, cuántos clics hay al día, cuántos megas ocupan y cuántos gigas son al año.

Resultado esperado:

Clics al día: 3,000,000
Clics: 660 MB al día
Clics: 240.9 GB al año

```python
visitas_al_dia = 250000
clics_totales_al_dia = visitas_al_dia * 12
bytes_por_clic = clics_totales_al_dia * 220
megas_totales = bytes_por_clic / 1000000
gigas_totales = (megas_totales / 1000) * 365
print(f"Clics al día: {clics_totales_al_dia:,} clics")
print(f"Clics: {megas_totales} MB al día")
print(f"Clics: {gigas_totales} GB al año")
```
